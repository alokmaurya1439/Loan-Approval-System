```python
#divide the data into input and output features
x = df.drop("loan_status",axis=1)
y = df.loan_status
```

```python
#split the data into training and testing datasets
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3,random_state=42)
```

# Model Pipeline

```python
#separte numeric and object column names
cat_cols = x_train.select_dtypes(include=['object', 'string']).columns.tolist()
num_cols = x_train.select_dtypes(include=['number']).columns.tolist()
```

```python
#preprocessing pipeline
preprocessor = ColumnTransformer([
                    ("encoder",OneHotEncoder(),cat_cols),
                    ("scaler",StandardScaler(),num_cols)
                 ]) 
```

# single model

```python
#model pipeline
pipeline = Pipeline([
        ("preprocessor",preprocessor),
        ("model",LogisticRegression(class_weight="balanced"))
    ])
pipeline
```

```python
#model pipeline training
pipeline.fit(x_train,y_train)
```

```python
#Training and test dataset efficiency
print("Training Eff: ",pipeline.score(x_train,y_train))
print("Testing Eff: ",pipeline.score(x_test,y_test))
```

```python
#predictions
ypred = pipeline.predict(x_test)
```

```python
confusion_matrix(y_test,ypred)
```

```python
print(classification_report(y_test,ypred))
```

```python
#new data
new_data =  x_test.sample()
pred = pipeline.predict(new_data)[0]
if pred==0:
  print("Loan rejected")
else:
  print("Loan approved")
```

# multi - model training

```python
models = {"lr":Pipeline([("preprocessor",ColumnTransformer([("encoder",OneHotEncoder(),cat_cols),
                                                                             ("scaler",StandardScaler(),num_cols)])),
                                          ("model",LogisticRegression(class_weight="balanced"))]),
          "svm":Pipeline([("preprocessor",ColumnTransformer([("encoder",OneHotEncoder(),cat_cols),
                                                                             ("scaler",StandardScaler(),num_cols)])),
                                          ("model",SVC())]),
          "DTC":Pipeline([("preprocessor",ColumnTransformer([("encoder",OneHotEncoder(),cat_cols),
                                                                             ("scaler",StandardScaler(),num_cols)])),
                                          ("model",DecisionTreeClassifier(max_depth=6))]),
          "RF":Pipeline([("preprocessor",ColumnTransformer([("encoder",OneHotEncoder(),cat_cols),
                                                                             ("scaler",StandardScaler(),num_cols)])),
                                          ("model",RandomForestClassifier(n_estimators=100,max_depth=6,random_state=42))])}
model_perf = {}
for model_name,pipeline in models.items():
  pipeline.fit(x_train,y_train)
  ypred = pipeline.predict(x_test)
  accuracy = accuracy_score(y_test,ypred)
  model_perf[model_name]=[accuracy,pipeline]
```

```python
perf=pd.Series(model_perf,name="accuracy")
perf=pd.DataFrame(perf).T
perf
```

```python
# save the dataframe into datasets
perf.to_csv("dataset/model_performance.csv")
```

```python
for a,b in model_perf.items():
  print(a,"--->",b[0])
```

```python
# fetching the best pipeline
best_model_name = max(model_perf.items(),key=lambda x:x[1])[0]
best_model = models[best_model_name]
best_model
```

```python
#predictions using best model
ypred = best_model.predict(x_test)
```

```python
confusion_matrix(y_test,ypred)
```

```python
sns.heatmap(confusion_matrix(y_test,ypred),annot=True,fmt="d",cmap="Blues")
```

```python
print(classification_report(y_test,ypred))
```

```python
#new data
new_data =  x_test.sample()
pred = best_model.predict(new_data)[0]
if pred==0:
  print("Loan rejected")
else:
  print("Loan approved")

new_data
```
